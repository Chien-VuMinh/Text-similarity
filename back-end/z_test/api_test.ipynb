```python
import httpx
url = "http://127.0.0.1:8000/api/sentences"

payload = {
    "sen1": "Tôi là Vũ minh chiến. Đây là một văn bản về. Hệ thống phát hiện tương đồng đa văn bản.",
    "sen2": "Tôi tên là minh Chiến. Hệ thống phát hiện văn bản.",
    "model": "gemini"
}

async with httpx.AsyncClient(timeout=60) as client:
    res = await client.post(url, json=payload)

    for content in res.json():
        print(content)
```
