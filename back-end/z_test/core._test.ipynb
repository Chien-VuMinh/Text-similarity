```python
a = """
Hệ thống, đồng đa văn bản? Hệ thống phát hiện !!!!Hệ thống phát hiện.
"""
```

```python
import re
lst = [s.strip() for s in re.split(r'(?<=[.?!])', a) if s.strip()]

original_sen = []
ind = 0
complete_sen = ''
while ind < len(lst):
    s = lst[ind]
    if len(s) > 1:
        if complete_sen:
            original_sen.append(complete_sen)
        complete_sen = s
    else:
        complete_sen += s
    ind += 1

original_sen.append(complete_sen)
original_sen
```

```python
import ollama

result = ollama.embed(
    model="bge-m3:567m",
    input=["hệ thống, đồng đa văn bản", "vu minh chien"]             
)
```

```python
import numpy as np
client = ollama.AsyncClient()

r = await client.embed(
    model="bge-m3:567m",
    input=["hệ thống, đồng đa văn bản", "vu minh chien"],
    dimensions=10           
)

r = np.array(r.embeddings)
```

```python
import asyncio
async def time1():
    await asyncio.sleep(1)
    await asyncio.sleep(2)
    print(1)

async def time2():
    t1 = asyncio.create_task(asyncio.sleep(1))
    t2 = asyncio.create_task(asyncio.sleep(2))
    print(1)
    await t1
    await t2

await time2()
```

```python
rows, cols = (np.array([0, 0, 0, 1, 1, 4]), np.array([1, 2, 3, 1, 3, 5]))

# Sử dụng một dictionary comprehension kết hợp với np.split
unique_rows, split_indices = np.unique(rows, return_index=True)
adj_dict = dict(zip(unique_rows, np.split(cols, split_indices[1:])))

print(adj_dict)
```
